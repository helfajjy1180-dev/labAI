```python
text = "Bonjour ! J'apprends le NLP, avec Python... C'est génial @2026."
print(text)
```

```python
text_lower = text.lower()
print(text_lower)
```

```python
text_clean = text_lower.translate(str.maketrans("", "", string.punctuation))
print(text_clean)
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text_clean, language="french")
print(tokens)
```

```python
from nltk.corpus import stopwords

stop_words = stopwords.words('french')
print(stop_words)
```

```python
filtered_words = [word for word in tokens if word not in stop_words]

print(filtered_words)
```

```python
stop_words.append("bonjour")

filtered_words = [word for word in tokens if word not in stop_words]

print(filtered_words)
```

```python
nombre_initial = len(tokens)
nombre_final = len(filtered_words)

proportion = (nombre_initial - nombre_final) / nombre_initial * 100

print("Nombre initial :", nombre_initial)
print("Nombre final :", nombre_final)
print("Proportion supprimée :", proportion, "%")
```
